# 5. Feature engineering

```python
group_cols = [
    'LocationAbbr',
    'Stratification1',
    'Stratification2'
]
Q03 = Q03.sort_values(group_cols + ['YearStart'])
Q03['Data_Value_prev'] = (
    Q03.groupby(group_cols)['Data_Value']
    .shift(1)
)

Q03[['LocationAbbr', 'Stratification1', 'YearStart', 'Data_Value', 'Data_Value_prev']].head(10)
```

```python
Q03 = Q03.dropna(subset=['Data_Value_prev'])
Q03.shape
```

# 6. Train / validation / test split

```python
X = Q03.drop(columns=['Data_Value'])
y = Q03['Data_Value']
X.shape, y.shape
```

```python
X_train = X[X['YearStart'] <= 2019]
X_val = X[X['YearStart'].isin([2020, 2021])]
X_test = X[X['YearStart'] == 2022]
y_train = y[X['YearStart'] <= 2019]
y_val = y[X['YearStart'].isin([2020, 2021])]
y_test = y[X['YearStart'] == 2022]

X_train.shape, X_val.shape, X_test.shape
```

```python
X_train.dtypes
```

# 7. Data preprocessing

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import mean_squared_error

categorical_cols = [
    'LocationAbbr',
    'Class',
    'Topic',
    'StratificationCategory1',
    'Stratification1',
    'StratificationCategory2',
    'Stratification2'
]

encoder = OneHotEncoder(handle_unknown='ignore')

X_train_cat = encoder.fit_transform(X_train[categorical_cols])
X_val_cat = encoder.transform(X_val[categorical_cols])
X_test_cat = encoder.transform(X_test[categorical_cols])

numeric_cols = ['YearStart', 'Data_Value_prev']

X_train_num = X_train[numeric_cols]
X_val_num = X_val[numeric_cols]
X_test_num = X_test[numeric_cols]
```

```python
from scipy.sparse import hstack

X_train_final = hstack([X_train_num, X_train_cat])
X_val_final = hstack([X_val_num, X_val_cat])
X_test_final = hstack([X_test_num, X_test_cat])
```

# 8. Models

## 8.1. Linear Regression

```python
lr = LinearRegression()

lr.fit(X_train_final, y_train)
y_val_pred = lr.predict(X_val_final)
lr_val_mse = mean_squared_error(y_val, y_val_pred)

print('Linear Regression Validation MSE:', lr_val_mse)
```

## 8.2. Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=32
)

rf.fit(X_train_final, y_train)
y_val_pred_rf = rf.predict(X_val_final)
rf_val_mse = mean_squared_error(y_val, y_val_pred_rf)

print("Random Forest Validation MSE:", rf_val_mse)
```

## 8.3. Gradient Boosting Regressor

```python
from sklearn.ensemble import GradientBoostingRegressor

gb = GradientBoostingRegressor(random_state=42)

gb.fit(X_train_final.toarray(), y_train)
y_val_pred_gb = gb.predict(X_val_final.toarray())
gb_val_mse = mean_squared_error(y_val, y_val_pred_gb)

print("Gradient Boosting Validation MSE:", gb_val_mse)
```

# 9. Model comparison

```python
results = {
    'Linear Regression': lr_val_mse,
    'Random Forest': rf_val_mse,
    'Gradien Boosting': gb_val_mse
}
for model, mse in results.items():
    print(f'{model}: {mse:.4f}')
```

Linear Regression performed best on the validation set

# 10. Final evaluation

```python
y_test_pred = lr.predict(X_test_final)
test_mse = mean_squared_error(y_test, y_test_pred)
print('Test MSE:', test_mse)
```

```python
rmse = np.sqrt(test_mse)
print('Test RMSE:', rmse)
```

```python
from sklearn.dummy import DummyRegressor

baseline = DummyRegressor(strategy='mean')
baseline.fit(X_train_final, y_train)
y_baseline = baseline.predict(X_test_final)
baseline_mse = mean_squared_error(y_test, y_baseline)

print('Baseline MSE:', baseline_mse)
print('Linear Regression MSE:', test_mse)
```

```python
improvement = (baseline_mse - test_mse) / baseline_mse * 100
print(f'Improvement: {improvement:.2f}%')
```

```python
# plot actual and predicted values
plt.scatter(y_test, y_test_pred)

plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Actual vs Predicted')

plt.show()
```

```python
# Calculate residuals
residuals = y_test - y_test_pred

plt.scatter(range(len(residuals)), residuals)

plt.axhline(0, color='purple')
plt.xlabel('Observation')
plt.ylabel('Residual')
plt.title('Residuals')

plt.show()
```

# 11. Conclusions

- Linear Regression achieved the lowest validation MSE.
- The final Test MSE was 4.27 and Test RMSE was approximately 2.07.
- The model reduced MSE by approximately 72% compared with the baseline.
- Most residuals were close to zero, indicating that predictions were generally close to the actual values.
